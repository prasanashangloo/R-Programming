```r
x = c(2,4,6,8,10,12)

excount = function(x){
  count = 0
  for (xval in x) {
    if (xval/2 > 3)
      count = count+1
  }
  print(count)
}

excount(x)

# Expected output:
# [1] 3
```

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}

# Expected output:
# [1] "child1 sweet1"
# [1] "child1 sweet2"
# [1] "child1 sweet3"
# [1] "child2 sweet1"
# [1] "child2 sweet2"
# [1] "child2 sweet3"
# [1] "child3 sweet1"
# [1] "child3 sweet2"
# [1] "child3 sweet3"
```

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}

# Expected output:
# [1] "coffee"
# [1] "lemonade"
```

```r
i = 1
while (i < 10) {
  print(i^2)
  i = i+2
}

# Expected output:
# [1] 1
# [1] 9
# [1] 25
# [1] 49
# [1] 81
```

```r
sumfunction = function(){
  sum = 0
  number = as.integer(readline(prompt="Please select any number less than 25: "))
  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the While Loop: ", sum))
}

sumfunction()

# Example run (input 22):
# Please select any number less than 25:  22
# [1] "The sum of numbers received from the While Loop:  94"
# (22+23+24+25 = 94)
```

```r
sumfunction_demo = function(start_number){
  sum = 0
  number = start_number
  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the While Loop: ", sum))
}
sumfunction_demo(22)
```

```r
i = 1
repeat{
  print( i^2 )
  i = i+2
  if ( i > 10 ) break
}

# Expected output:
# [1] 1
# [1] 9
# [1] 25
# [1] 49
# [1] 81
```

```r
i = 1
repeat{
  i = i+1
  if (i < 10) next
  print(i^2)
  if (i >= 13) break
}

# Expected output:
# [1] 100
# [1] 121
# [1] 144
# [1] 169
```

```r
abc = function(x){
  x^2
}

abc(3)
abc(6)
abc(9)

# Expected output:
# [1] 9
# [1] 36
# [1] 81
```

```r
abc = function(x,y){
  x^2+y^2
}

abc(3,4)
abc(10, 10)
abc(-2, -3)

# Expected output:
# [1] 25
# [1] 200
# [1] 13
```

```r
abc = function(x){
  sin(x)^2+cos(x)^2 + x
}

abc(9)
abc(99)
abc(-15)

# Expected output:
# [1] 10
# [1] 100
# [1] -14
```

```r
abc = function() {
  for(i in 1:3) {
    print(i^3)
  }
}

abc()

# Expected output:
# [1] 1
# [1] 8
# [1] 27
```

```r
seq(from=2, to=4)
seq(from=4, to=2)
seq(from=-4, to=4)

# Expected output:
# [1] 2 3 4
# [1] 4 3 2
# [1] -4 -3 -2 -1  0  1  2  3  4
```

```r
seq(from=10, to=20, by=2)

# Expected output:
# [1] 10 12 14 16 18 20
```

```r
seq(from=20, to=10, by=-2)

# Expected output:
# [1] 20 18 16 14 12 10
```

```r
seq(from=3, to=-2, by=-0.5)

# Expected output:
# [1] 3.0 2.5 2.0 1.5 1.0 0.5 0.0 -0.5 -1.0 -1.5 -2.0
```
