```r
# --------------------- Lecture 28: Repeats ---------------------
help("rep")
```

```r
# Repeat an object n-times
rep(3.5, times=10)
rep(1:4, 2)
```

```r
# Repeat each cell n-times
x = 1:4
x
rep(x, times = 3)
rep(x, each = 3)
```

```r
# Ordering of each and times
rep(1:4, each = 2)
rep(1:4, each = 2, times = 3)
rep(1:4, times = 3, each = 2)
```

```r
# Every object repeated a different number of times
rep(1:4, 2:5)
```

```r
ans = seq(from=2, to=8, by=2)
ans
rep(1:4, ans)
```

```r
# rep applied to a matrix
x = matrix(nrow=2, ncol=2, data=1:4, byrow=T)
x
rep(x, 3)
```

```r
# Repetition of characters
rep(c("a", "b", "c"), 2)
rep(c("apple", "banana", "cake"), 2)
```

```r
# Repetition of characters for pre-specified length
rep(2, length.out=5)
rep(2, length=5)
rep(c(2,3), length=5)
rep(c(2,3,4), length=5)
```

```r
rep("apple", length=5)
rep(c("a", "b", "c"), length=2)
rep(c("a", "b", "c"), length=5)
```

```r
# --------------------- Lecture 29: Sorting, Ordering and Mode ---------------------
# Sorting
y = c(8,5,7,6)
y
sort(y)
sort(y, decreasing = TRUE)
```

```r
# Ordering
y = c(9,8,5,7,6)
y
order(y)
order(y, decreasing = TRUE)
```

```r
# Mode
mode(2.432)
mode(c(3,4,5,6,7,8))
mode("India")
mode(c("India", "CANADA"))
```

```r
mode(factor(c("UP", "MP")))
mode(list("India", "USA"))
mode(data.frame(x=1:2, y=c("India", "USA")))
mode(print)
```

```r
# List with different object types
z1 = list( c("water", "juice", "lemonade"),
           rep(1:4, each=2),
           matrix(data=5:8, nrow=2, ncol=2, byrow=T) )
z1
```

```r
# Accessing elements of a list
z1[[1]]
z1[1][2]     # returns NULL
z1[[1]][2]   # returns "juice"
```
