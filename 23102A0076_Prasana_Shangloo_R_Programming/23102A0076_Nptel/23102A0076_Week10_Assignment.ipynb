```r
# sub() replaces the first instance
y = "Number of participants: 25"

sub("25", "30", y)
```

```r
# sub() replaces only the first matching instance
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."

y
sub("Mr. Singh", "Professor Jha", y)
```

```r
# gsub() replaces all matching instances
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."

gsub("Mr. Singh", "Professor Jha", y)
```

```r
# grep() searches for matches and returns indices by default
str = c(
  "R Course",
  "exercises",
  "include examples of R language"
)

grep("ex", str, value = TRUE)
grep("ex", str, value = FALSE)
```

```r
# ignore.case controls case sensitivity
str = c(
  "R Course",
  "exercises",
  "include examples of r language",
  "in R software."
)

grep("R", str, ignore.case = FALSE, value = TRUE)
grep("R", str, ignore.case = TRUE, value = TRUE)

grep("R", str, ignore.case = TRUE, value = FALSE)
grep("R", str, ignore.case = FALSE, value = FALSE)
```

```r
# grep() example
x = "R course 24.07.2021"
y = "Number of participants: 25"

c(x, y)

grep("our", c(x, y))
grep("Num", c(x, y))
```

```r
# grepl() returns TRUE/FALSE for each element
str = c(
  "R Course",
  "exercises",
  "include examples of R language"
)

grepl("R", str)
grepl("ex", str)
```

```r
# Create two data frames
df1 = data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)

df2 = data.frame(
  state = c("UP", "MP", "AP", "JK"),
  samplesize = c(100, 200, 300, 400),
  surveycompleted = c("Yes", "No", "Yes", "No")
)

df1
df2
```

```r
# cbind() combines columns side-by-side
cbind(df1, df2)
```

```r
# merge() joins data frames using the common column
merge(df1, df2, by = "state")
```

```r
# rbind() stacks rows vertically
df11 = data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)

df22 = data.frame(
  state = c("Bihar", "Delhi", "Punjab"),
  popnsize = c(100, 200, 300)
)

df11
df22
```
