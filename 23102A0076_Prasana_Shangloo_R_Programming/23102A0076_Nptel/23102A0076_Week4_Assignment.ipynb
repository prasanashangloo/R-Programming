```r
x = 8
(x < 10) || (x < 2)   # single value -> TRUE
```

```r
x = 18
(x < 10) || (x < 2)   # FALSE
```

```r
x = c(8, 18)
(x < 10) || (x < 2)   # || only checks the FIRST element
```

```r
(x < 10) | (x < 2)    # | operates element-wise on the whole vector
```

```r
x = 5
(x < 10) && (x > 2)   # TRUE
```

```r
x = 15
(x < 10) && (x > 2)   # FALSE
```

```r
x = c(8, 18)
(x < 10) && (x > 2)   # only checks first element
```

```r
(x < 10) & (x > 2)    # operates on all elements
```

```r
x = 1:6
x
```

```r
(x > 2) & (x < 5)          # element-wise logical AND
```

```r
x[(x > 2) & (x < 5)]       # values between 2 and 5
```

```r
(x > 2) | (x < 5)          # element-wise logical OR
```

```r
x[(x > 2) | (x < 5)]
```

```r
x = 1:6
(x > 2) && (x < 5)     # equivalent to using only x[1]
```

```r
(x[1] > 2) & (x[1] < 5)   # same result as above
```
