```r
c(2,3,5,7) + c(-2,-3,-5,8)
# 2+(-2), 3+(-3), 5+(-5), 7+8
```

```r
2^3     # Command for power operator
2**3    # Same result
```

```r
2^0.5    # square root via power
2**0.5
2^-0.5   # negative power -> reciprocal of root
```

```r
c(2,3,5,7)^2   # 2^2, 3^2, 5^2, 7^2
```

```r
c(2,3,5,7)^c(2,3)   # 2^2, 3^3, 5^2, 7^3 -> recycling happens since exponent vector is shorter
```

```r
c(1,2,3,4,5,6)^c(2,3,4)   # exponent vector recycled: 1^2,2^3,3^4,4^2,5^3,6^4
```

```r
c(2,3,5,7)^c(2,3,4)   # length 4 is not a multiple of length 3 -> warning expected
```

```r
2 %/% 2
5 %/% 2
7 %/% 3
```

```r
c(2,3,5,7) %/% 2   # scalar divisor applied to each element
```

```r
c(2,3,5,7) %/% c(2,3)   # vector divisor, recycled: 2/%2, 3/%3, 5/%2, 7/%3
```

```r
c(2,3,5) %/% c(2,3)   # length 3 not a multiple of length 2 -> warning expected
```

```r
2 %% 2
3 %% 2
7 %% 3
7 %% 4
```

```r
c(2,3,5,7) %% 2
```

```r
c(2,3,5,7) %% c(2,3)   # recycled
```

```r
c(2,3,5) %% c(2,3)   # warning: lengths not multiples
```

```r
max(1.2, 3.4, -7.8)
max(c(1.2, 3.4, -7.8))   # same result for max
```

```r
min(1.2, 3.4, -7.8)
min(c(1.2, 3.4, -7.8))   # same result for min
```
