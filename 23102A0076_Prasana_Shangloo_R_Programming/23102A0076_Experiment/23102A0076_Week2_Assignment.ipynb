```r
c(2,3,5,7) + c(-2,-3,-5,8)
# element-wise: 2+(-2), 3+(-3), 5+(-5), 7+8
```

```r
c(2,3,5,7) + c(8,9)  # !!! ATTENTION: recycling
# 2+8, 3+9, 5+8, 7+9  (c(8,9) is recycled as c(8,9,8,9))
```

```r
c(2,3,5,7) + c(8,9,10)  # error/warning message
# 2+8, 3+9, 5+10, 7+8  (recycling with leftover element triggers a warning)
```

```r
c(2,3,5,7) - c(-2,-3,-5,8)
# 2-(-2), 3-(-3), 5-(-5), 7-8
```

```r
c(12,13,15,17) - c(8,9)  # !!! ATTENTION: recycling
# 12-8, 13-9, 15-8, 17-9
```

```r
c(12,13,15,17) - c(8,9,10)  # warning: length mismatch
# 12-8, 13-9, 15-10, 17-8
```

```r
c(2,3,5,7) * c(-2,-3,-5,8)
# 2*(-2), 3*(-3), 5*(-5), 7*8
```

```r
c(2,3,5,7) * c(8,9)  # !!! ATTENTION: recycling
# 2*8, 3*9, 5*8, 7*9
```

```r
c(2,3,5,7) * c(8,9,10)  # warning: length mismatch
# 2*8, 3*9, 5*10, 7*8
```

```r
c(24,20,8,16) / c(3,4,2,8)
# 24/3, 20/4, 8/2, 16/8
```

```r
c(24,20,8,16) / c(4,2)  # !!! ATTENTION: recycling
# 24/4, 20/2, 8/4, 16/2
```

```r
c(24,20,8,16) / c(4,2,8)  # warning: length mismatch
# 24/4, 20/2, 8/8, 16/4
```

```r
# number -> character
x = 20
is.numeric(x)

y = as.character(x)
is.numeric(y)
is.character(y)
y
```

```r
# character -> number (works only if the string looks like a number)
y = "apple"
is.numeric(y)
is.character(y)

z = as.numeric(y)   # produces NA with a warning: NAs introduced by coercion
is.numeric(z)
is.character(z)
z
```
